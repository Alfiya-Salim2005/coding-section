```python
n=int(input("Enter the number of monkeys"))
k=int(input("Enter the number of bananas eatable by single monkey"))
j=int(input("Enter the number of peanuts eatable by single monkey"))
m=int(input("Enter the number of bananas"))
p=int(input("Enter the number of peanuts"))
if(k==0 or j==0):
  print("invalid input")
else:
  left=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    left=left-1
  print("Number of monkeys in tree=",left)
```

```python
n = int(input("Enter the number of elements: "))
arr = []
for i in range(n):
    arr.append(int(input("Enter the element: ")))
ranks = list(map(int, input().split()))
current_min = ranks[0]
cuts = 0
for rank in ranks[1:]:
    if rank < current_min:
        current_min = rank
        cuts += 1
print(cuts)
```
