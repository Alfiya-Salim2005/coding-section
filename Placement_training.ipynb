```python
n=int(input("Enter the number of monkeys"))
k=int(input("Enter the number of bananas eatable by single monkey"))
j=int(input("Enter the number of peanuts eatable by single monkey"))
m=int(input("Enter the number of bananas"))
p=int(input("Enter the number of peanuts"))
if(k==0 or j==0):
  print("invalid input")
else:
  left=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    left=left-1
  print("Number of monkeys in tree=",left)
```

```python
n=int(input("Enter max no. of candies:"))
k=int(input("Enter min no. of candies:"))
con=1
r=n
while(con==1):
  m=int(input("Enter amout of candies needed:"))
  if(m>n):
    print("invalid")
  else:
    r=r-m
    if(r<=k):
      r=n
    print("Sold=",m)
    print("Remaining=",r)
    con=int(input("press 0 to exit and 1 to continue:"))
```
